```python
#Filtering of data
# print(df[df["aqi"]>100]) # this will give all the rows where the aqi is greater than 100
print(df[df["country"]=="US"]) # this will give all the rows where the country is US
print(df[(df["country"]=="US") & (df["aqi"]>100)]) # this will give all the rows where the country is US and the aqi is greater than 100
df_cpy=df[(df["country"]=="US") & (df["aqi"]>100)&(df["city"]=="New York") ][["city","aqi"]]
df_cpy.iloc[1]
# df_cpy.loc[1] # this will give error because the index of the filtered dataframe is not reset, so we need to reset the index before using loc
df_cpy.loc[3] # this will give the same result as iloc because we are selecting the first row of the filtered dataframe which has index 3 in the original dataframe
```

```python
# filtering of data using query method
df.query("aqi>100") # this will give all the rows where the aqi is greater than 100
aqi_val=100
df.query("aqi>@aqi_val") # this will give all the rows where the
df.query("country=='US' & aqi>100 & city=='New York'& temperature>30") # this will give all the rows where the country is US and the aqi is greater than 100
#this will return a copy of the original dataframe with the filtered data, so any changes made to the filtered dataframe will not affect the original dataframe.
```

```python
#cleaning of data
data=pd.read_csv("raw_data.csv")
data.isnull()
age_mean=data["age"].mean()
cleaned_df=data.copy()
cleaned_df.fillna(age_mean,inplace=True)
cleaned_df.duplicated()
cleaned_df.drop_duplicates(inplace=True)
cleaned_df
```

```python
# cleaned_df["tax"]=cleaned_df["income"].apply(lambda x: "20%" if x>60000 else "10%")
# gender={"Male":"M","Female":"F","Unknown":"O"}   
# cleaned_df["gender"]=cleaned_df["gender"].map(gender)
# cleaned_df=cleaned_df.assign(newincome=cleaned_df["income"]*1.1)
# cleaned_df[ "country" ]=cleaned_df[ "country" ].replace("USA", "us")
cleaned_df
```

```python
cleaned_df.to_csv("cleaned_data.csv", index=False)
```

```python
cleaned_df.groupby("country")["income"].mean() # this will give the average income for each country
cleaned_df.groupby("country")["income"].min()
cleaned_df.groupby("country")["income"].max()
cleaned_df.groupby("gender")["income"].min()
cleaned_df.groupby("gender")["income"].max()
cleaned_df.groupby("country")["income"].agg(["max","min","mean"]) # this will give the max, min and mean income for each country
cleaned_df.groupby("country").agg({
    "income":  "mean",
    "age":"mean"
})
cleaned_df.groupby("country").agg(
    max_income=("income",  "max"),
    mean_age=("age","mean")
)
```

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
pd.merge(df_customers, df_orders, on="customer_id") 
pd.merge(df_customers, df_orders, on="customer_id" ,how="left") 
pd.merge(df_customers, df_orders, on="customer_id",how="right") 
pd.merge(df_customers, df_orders, on="customer_id",how="outer") 
```
