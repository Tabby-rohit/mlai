```python
import json
```

```python
def load_data(filename) :
   with open(filename,"r") as f:
       data=json.load(f)
   return data    
```

```python
data=load_data("store_data.json")
print(f"data={data}")
```

```python
def clean_data(data):
    cleaned_data=[]
    unique_user=set()
    for user in data:
        text_to_num={"one":1,"two":2,'three':3,"four":4,"five":5,"six":6,"seven":7,"eight":8,"nine":9,"ten":10}
        raw_rating=str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[ raw_rating]
        user["rating"]=raw_rating
        raw_age=user.get("age")
        #cleaning age
        if raw_age==None:
            user["age"]=raw_age

        #deduplication
        #print(user)
        
        #print(user['name'])
        
        
        if(user['name'].strip() in unique_user):
            continue
        cleaned_data.append(user)
        unique_user.add(user['name'].strip())
        

    return cleaned_data
        
            
    
```

```python
data=clean_data(data)
```

```python
sum=0
i=0
j=0
perc_rating_less_than_3=100
for user in clean_data(data):
    sum+=float(user["rating"])
    i+=1
    if float(user["rating"])<=3:
        j+=1
    
    
avg_rating=sum/i
perc_rating_less_than_3*=j/i
print(perc_rating_less_than_3)
print(avg_rating)
```

```python
#recommendation
def get_recommendation(data):
    recommendations=[]
    for user in data:
        curr_recomm={}
        if float(user['rating'])>=4 :
            curr_recomm['brand']="apple"
        else:
            curr_recomm['brand']="samsung"
        curr_recomm['name']=user["name"]
        recommendations.append(curr_recomm)
    return recommendations 
    
```

```python
get_recommendation(data)
```
