```python
# histograms
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)
data = np.random.normal(50,10,100)
plt.hist(data,color="skyblue",edgecolor="black",bins=20) #histogram
# plt.show()
```

```python
#boxplot
data= [7, 8, 5, 6, 9, 10, 4, 15, 12]
plt.boxplot(data,whis=1.2) #boxplot
# plt.grid(True)
plt.show()
```

```python
#stack plot
x = [1, 2, 3, 4, 5,6]
y1 = [2, 3, 4, 2, 6,10]    
y2 = [1, 2, 3, 0, 5,100]
plt.stackplot(x, y1, y2, labels=['y1', 'y2'], colors=['skyblue', 'lightgreen']) #stackplot
plt.legend(loc='upper left')
```

```python
plt.subplot(2,1,1) #subplot
plt.plot([1, 2, 3], [4, 5, 6])
plt.subplot(2,2,2)
plt.stackplot(x, y1, y2, labels=['y1', 'y2'], colors=['skyblue', 'lightgreen'])
```

```python
fig, ax = plt.subplots(2,2,figsize=(10,10))
ax[0,0].plot([1, 2, 3], [4, 5, 6])
ax[0,0].set_title("Line Plot")
ax[0,1].stackplot(x, y1, y2, labels=['y1', 'y2'], colors=['skyblue', 'lightgreen'])
ax[0,1].set_title("Stack Plot")
ax[1,0].hist(data,color="skyblue",edgecolor="black",bins=20)
ax[1,0].set_title("Histogram")
ax[1,1].boxplot(data,whis=1.2)
ax[1,1].set_title("Box Plot")
fig.suptitle("Data Visualization Examples")
plt.tight_layout()
```
